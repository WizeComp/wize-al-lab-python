```python
# STEP 1: Question
question = "I am predicting a penguin's species from its flipper length, bill length, bill depth, and body mass."
print(question)

# STEP 2: Data
data = load("penguins.csv").dropna()
print(data.shape)
print("Missing values:", data.isna().sum().sum())
data.head()
```

```python
# STEP 3: Explore
for species, group in data.groupby("species"):
    plt.scatter(group["flipper_length_mm"], group["bill_length_mm"], label=species)
plt.title("Flipper vs bill length by species"); plt.xlabel("Flipper length (mm)"); plt.ylabel("Bill length (mm)"); plt.legend()
plt.show()
```

```python
# STEP 4 and 5: Model and Evaluate
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix

features = ["flipper_length_mm", "bill_length_mm", "bill_depth_mm", "body_mass_g"]
label = "species"
X_train, X_test, y_train, y_test = train_test_split(data[features], data[label], test_size=0.25, random_state=42)

model_a = KNeighborsClassifier(n_neighbors=3).fit(X_train, y_train)
model_b = DecisionTreeClassifier(max_depth=3, random_state=42).fit(X_train, y_train)

acc_a = model_a.score(X_test, y_test) * 100
acc_b = model_b.score(X_test, y_test) * 100
print("Model A (KNN) accuracy:", round(acc_a, 1), "%")
print("Model B (tree) accuracy:", round(acc_b, 1), "%")
best_model = model_a if acc_a >= acc_b else model_b
print("Chosen model:", type(best_model).__name__)
print("Confusion matrix (rows = real, columns = predicted):")
print(confusion_matrix(y_test, best_model.predict(X_test)))
```

```python
# STEP 6: Limits
limits = """
The penguins come from three islands in one region of Antarctica, so the model may not work for penguins elsewhere or for other species.
If a wildlife app trusted this model blindly, it would confidently mislabel any penguin it has never seen, so a human should check surprising results.
"""
print(limits)

# SPICY: live demo
def predict_one():
    row = {}
    for f in features:
        row[f] = float(input(f"Enter {f}: "))
    prediction = best_model.predict(pd.DataFrame([row]))[0]
    print("The model predicts:", prediction)

# Uncomment to run live during the showcase:
# predict_one()

# Colab bonus: a Gradio web app in five lines
# import gradio as gr
# def app(flipper, bill_len, bill_dep, mass):
#     return best_model.predict(pd.DataFrame([[flipper, bill_len, bill_dep, mass]], columns=features))[0]
# gr.Interface(fn=app, inputs=["number", "number", "number", "number"], outputs="text").launch()
```
