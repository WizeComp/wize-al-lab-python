## Hook: Which is true?

In a class of 120 students, do the students who sleep more have higher focus scores? Or is that a myth? Guess first, then let the data decide.

```python
# Setup: run this cell first.
import pandas as pd
import matplotlib.pyplot as plt

# If a data file is not found next to this notebook (for example on Google Colab),
# it is loaded from the Wize data folder online instead. Replace this URL after publishing.
DATA_URL = "https://raw.githubusercontent.com/wizeacademy/ml-ai-6-8/main/notebooks/data/"

def load(name):
    """Load a Wize dataset by file name, from the local data folder or from the web."""
    try:
        return pd.read_csv("data/" + name)
    except Exception:
        return pd.read_csv(DATA_URL + name)

print("Setup complete. pandas and matplotlib are ready.")
```

## Teach 1: Looking at a dataset

`pandas` stores a table as a **DataFrame**. These are the first four commands every data scientist runs.

```python
students = load("wize_students.csv")

print("Rows and columns:", students.shape)
print("Column names:", list(students.columns))
students.head()
```

```python
# describe() gives quick statistics for every number column
students.describe()
```

## Teach 2: Asking questions with code

Filtering keeps only the rows that match a condition. `value_counts()` counts categories. `groupby()` splits the table into groups and computes something for each group.

**Concept checkpoint:** before you run the next cell, predict which grade has the most students.

```python
# Filtering: students who sleep 9 hours or more
well_rested = students[students["hours_sleep"] >= 9]
print("Well rested students:", len(well_rested))

# Counting categories
print(students["favorite_subject"].value_counts())

# Group by grade and take the average of every number column
print(students.groupby("grade")[["hours_sleep", "screen_hours", "focus_score"]].mean().round(1))
```

## Teach 3: Charts with matplotlib

A picture is the fastest way to spot a pattern. Three charts cover most questions:

- **Bar chart** for categories
- **Histogram** for one number (how is it spread out?)
- **Scatter plot** for two numbers (do they move together?)

```python
counts = students["favorite_subject"].value_counts()
plt.bar(counts.index, counts.values, color="#2271b1")
plt.title("Favorite subject")
plt.xlabel("Subject")
plt.ylabel("Number of students")
plt.xticks(rotation=30)
plt.show()

plt.scatter(students["hours_sleep"], students["focus_score"], color="#f7941d")
plt.title("Sleep vs focus score")
plt.xlabel("Hours of sleep")
plt.ylabel("Focus score")
plt.show()
```
