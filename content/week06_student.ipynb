## Teach 1: A sentiment score

**Sentiment analysis** guesses the feeling in a piece of text. VADER uses a giant list of words with scores (great = +3.1, awful = -3.1) plus rules for CAPITALS, exclamation marks, and words like "not". The **compound** score runs from -1 (very negative) to +1 (very positive).

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()

sentences = [
    "This game is amazing!!!",
    "This game is fine I guess.",
    "This game is terrible and it crashed.",
    "This game is not bad at all.",
]
for s in sentences:
    score = analyzer.polarity_scores(s)["compound"]
    print(round(score, 2), "  ", s)
```

## Teach 2: Where word lists fail

**Concept checkpoint:** predict the score for "Oh great, another crash." Then run it. Sarcasm, slang, and context are hard for a word list. Every AI has **limits**, and the honest data scientist finds them on purpose.

```python
tricky = ["Oh great, another crash.", "It is sick, I love it", "Not the worst game ever"]
for s in tricky:
    print(round(analyzer.polarity_scores(s)["compound"], 2), "  ", s)
```
