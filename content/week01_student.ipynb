## Teach 1: A rule based classifier

A **classifier** puts things into categories. The simplest classifier is a set of `if` rules written by a human. Read the code, **predict what it prints**, then run it.

```python
def classify_pet(purrs, fetches):
    """Return 'cat' or 'dog' using hand written rules."""
    if purrs:
        return "cat"
    elif fetches:
        return "dog"
    else:
        return "not sure"

print(classify_pet(purrs=True,  fetches=False))
print(classify_pet(purrs=False, fetches=True))
print(classify_pet(purrs=False, fetches=False))
```

## Teach 2: Rules break. Data scales.

What about a cat that fetches? A parrot? A rock? Every new case needs a new rule. **Machine learning** flips this: instead of writing rules, we give the computer **examples** (data with labels) and it finds the pattern itself.

A dataset is a table. Each **row** is one example. Each **column** is a **feature** (something we measured). One special column is the **label** (the answer we want to predict).

**Concept checkpoint:** in the table below, which column is the label?
