## Teach 1: Calling an AI from Python

Our program builds a web address with the question attached, sends it, and reads the answer back. The `ask_ai` function hides the details so you can focus on the prompt.

```python
import os, urllib.parse, requests

def ask_ai(prompt):
    key = "POLLINATIONS_KEY"
    url = "https://gen.pollinations.ai/text/" + urllib.parse.quote(prompt)
    r = requests.get(url, params={"key": key}, timeout=60)
    r.raise_for_status()
    return r.text

print(ask_ai("Say hello to a middle school coding class in one sentence."))
```

## Teach 2: Prompts are instructions, personas are prompts

The model does exactly what the words ask, so **clear prompts get clear answers**. A **persona** is a prompt that sets the character before the user's question.

**Concept checkpoint:** which prompt will get a better answer: "penguins" or "Explain in two sentences why penguins cannot fly, for a 12 year old"?

```python
persona = "You are a cheerful robot tour guide at an aquarium. Answer in two short sentences. "
question = "Why can't penguins fly?"
print(ask_ai(persona + question))
```

## Teach 3: The model is not always right

LLMs predict likely words; they do not look things up. That means they can **hallucinate**: state something false with total confidence. Run the cell, then check the answer against a source you trust.

```python
print(ask_ai("In one sentence, who won the FIFA World Cup in 2030?"))
print()
print(ask_ai("How many letters are in the word 'strawberry'? Answer with just the number."))
```
